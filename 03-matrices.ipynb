## 3.2 Where the two arrows land

Two columns tell you everything about a 2 by 2 matrix: where the two unit arrows land. You read the common moves, stretch, rotate, shear, flip and squash, straight from them.

[Read the lesson](https://aicourses.dev/mathematics/matrices/where-the-arrows-land)

### Reading a matrix as two landing spots

The same thing in NumPy. The symbol `@` multiplies a matrix by a vector, and the next lesson shows exactly how; here you only need what it gives:

````python
import numpy as np

A = np.array([[1, -1],
              [1,  1]])      # columns (1, 1) and (-1, 1)
right, up = np.array([1, 0]), np.array([0, 1])
print(A @ right, A @ up)     # where the two arrows land
print(A[:, 0], A[:, 1])      # the two columns of A
print(A @ np.array([3, 1]))  # where the point (3, 1) lands
````

*The lesson shows:*

```text
[1 1] [-1  1]
[1 1] [-1  1]
[2 4]
```

### Five moves as five pairs of columns

The slant keeps the right arrow where it is and pushes the up arrow sideways to (1, 1), so the unit square leans into a parallelogram. The squash is the matrix with columns (1, 2) and (2, 4): both arrows lie on the same line, so everything does. Take the point (2, 3) through three of them in NumPy:

````python
shear = np.array([[1, 1], [0, 1]])
flip = np.array([[-1, 0], [0, 1]])
squash = np.array([[1, 2], [2, 4]])
point = np.array([2, 3])
for name, M in [("shear", shear), ("flip", flip), ("squash", squash)]:
    print(name, M @ point)
````

*The lesson shows:*

```text
shear [5 3]
flip [-2  3]
squash [ 8 16]
```

## 3.3 Matrix times vector

A matrix times a vector is the one calculation a neural network does most. Here you see it two ways, as a weighted sum of columns and as one dot product per row.

[Read the lesson](https://aicourses.dev/mathematics/matrices/matrix-times-vector)

### In NumPy

First the two readings for the 2 by 2 example, written out with loops, then the one-line version:

````python
import numpy as np

A = np.array([[2, 1], [1, 3]])
x = np.array([1, 2])

print(x[0] * A[:, 0] + x[1] * A[:, 1])   # a weighted sum of columns
rows = []
for i in range(2):                       # one dot product per row
    total = 0
    for j in range(2):
        total += A[i, j] * x[j]
    rows.append(total)
print(np.array(rows))
print(A @ x)                             # the same, in one line
````

*The lesson shows:*

```text
[4 7]
[4 7]
[4 7]
```

The 3 by 2 matrix, with its shapes:

````python
W = np.array([[1, 2], [0, 1], [-1, 1]])
print(W.shape, (W @ np.array([2, 1])).shape)
print(W @ np.array([2, 1]))
print(W @ np.array([3, -1]))
````

*The lesson shows:*

```text
(3, 2) (3,)
[ 4  1 -1]
[ 1 -1 -4]
```

## 3.4 Machines in a row: matrix multiplication

Two machines in a row make one machine, and order matters. The lesson also shows why two layers with no bend between them collapse into one, which explains why networks need bends.

[Read the lesson](https://aicourses.dev/mathematics/matrices/matrix-multiplication)

### In NumPy

The sum over $k$ as two loops, and then `@`:

````python
import numpy as np

R = np.array([[0, -1], [1, 0]])    # quarter turn
S = np.array([[1, 1], [0, 1]])     # slant
P = np.zeros((2, 2), dtype=int)
for i in range(2):
    for j in range(2):
        P[i, j] = R[i] @ S[:, j]   # row i of R with column j of S
print(P)
print(R @ S)
print(S @ R)
print(R * S)                       # entry by entry: a different thing
````

*The lesson shows:*

```text
[[ 0 -1]
 [ 1  1]]
[[ 0 -1]
 [ 1  1]]
[[ 1 -1]
 [ 1  0]]
[[ 0 -1]
 [ 0  0]]
```

The first two are the same, and the third is $\mathbf{S}\mathbf{R}$. The last line, with `*`, multiplies matching entries and is not a chain of machines: it is a common slip, so keep `@` for chaining. Now the two layers:

````python
W1 = np.array([[1, 2], [0, 1], [-1, 1]])   # 2 in, 3 out
W2 = np.array([[1, 0, 1], [2, 1, 0]])      # 3 in, 2 out
x = np.array([2, 1])
print(W2 @ (W1 @ x))
M = W2 @ W1
print(M)
print(M @ x)
print(W1.size + W2.size, M.size)
````

*The lesson shows:*

```text
[3 9]
[[0 3]
 [2 5]]
[3 9]
12 4
```

## 3.5 Rotations, and other moves that keep lengths

Some matrices turn space without stretching it, and rotations are the simplest. They are built from the sine and cosine of Module 01 and return as rotary positions in Module 12.

[Read the lesson](https://aicourses.dev/mathematics/matrices/rotations)

### In NumPy

The matrix as a function of the angle, in radians (a half turn is $\pi$ radians, so 45 degrees is $\pi / 4$):

````python
import numpy as np

def rot(angle):
    c, s = np.cos(angle), np.sin(angle)
    return np.array([[c, -s], [s, c]])

R = rot(np.pi / 4)
x = np.array([3, 4])
print(R.round(4))
print((R @ x).round(4), np.linalg.norm(x), np.linalg.norm(R @ x))
print(np.allclose(R.T @ R, np.eye(2)))
````

*The lesson shows:*

```text
[[ 0.7071 -0.7071]
 [ 0.7071  0.7071]]
[-0.7071  4.9497] 5.0 5.0
True
```

### Where this lives in AI

Rotary position embedding rotates each pair of coordinates of a query or a key by an angle, the position times a rate that is different for each pair. In the paper the rate for pair $i$ is $\theta_i = 10000^{-2(i-1)/d}$, for $i = 1, \dots, d/2$, in a vector of $d$ numbers, and the full rotation is a table with one 2 by 2 rotation for each pair and zeros elsewhere. The paper notes that the table is orthogonal, so by this lesson the lengths of the vectors are untouched. Its main property is one you can check with a single pair of coordinates. Rotate a query by its position $m$ and a key by its position $n$, using one made-up rate, and the dot product depends only on the difference $n - m$:

````python
q, k = np.array([1.0, 2.0]), np.array([3.0, 1.0])
theta = 0.5                      # a made-up rate, one pair of coordinates
for m, n in [(2, 5), (7, 10), (0, 3), (4, 5)]:
    print(m, n, round((rot(m * theta) @ q) @ (rot(n * theta) @ k), 4))
````

*The lesson shows:*

```text
2 5 5.3412
7 10 5.3412
0 3 5.3412
4 5 6.785
```

## 3.6 The determinant: how much area scales

The determinant is one number that says how much a machine scales area, and zero means it squashes space flat. That sets up the next lesson on what can be undone.

[Read the lesson](https://aicourses.dev/mathematics/matrices/the-determinant)

### In NumPy

Compute the area of the unit square after the move by the corners, with the shoelace rule for a polygon’s area, and then ask NumPy for the determinant:

````python
import numpy as np

A = np.array([[2, 1], [1, 3]])
square = np.array([[0, 1, 1, 0], [0, 0, 1, 1]])   # corners as columns

def area(P):
    x, y = P
    return 0.5 * abs(np.sum(x * np.roll(y, -1) - np.roll(x, -1) * y))

print(area(square), area(A @ square))
print(round(np.linalg.det(A), 4))
````

*The lesson shows:*

```text
1.0 5.0
5.0
```

The area goes from 1 to 5, and `np.linalg.det` gives the same 5. It works with floating point numbers, so it can return 5.000000000000001; rounding shows the value you mean. Now the other matrices, including a chain:

````python
stretch = np.array([[3, 0], [0, 2]])
flip = np.array([[-1, 0], [0, 1]])
squash = np.array([[1, 2], [2, 4]])
for name, M in [("stretch", stretch), ("flip", flip), ("squash", squash),
                ("A swapped", A[:, ::-1])]:
    print(name, round(np.linalg.det(M), 4))
print(round(np.linalg.det(A @ stretch), 4))
````

*The lesson shows:*

```text
stretch 6.0
flip -1.0
squash 0.0
A swapped -5.0
30.0
```

## 3.7 Undoing a matrix: inverse and rank

Some moves can be undone and some cannot, and rank counts how many directions survive. This is the idea behind compressing a matrix in Module 10.

[Read the lesson](https://aicourses.dev/mathematics/matrices/inverse-and-rank)

### In NumPy

Solve, invert and check the rank of the example:

````python
import numpy as np

A = np.array([[2, 1], [1, 3]])
b = np.array([3, 5])
x = np.linalg.solve(A, b)
print(x, A @ x)
print(np.linalg.inv(A))
B = np.array([[1, 2], [2, 4]])
print(np.linalg.matrix_rank(A), np.linalg.matrix_rank(B))
print(B @ np.array([3, 0]), B @ np.array([1, 1]))
````

*The lesson shows:*

```text
[0.8 1.4] [3. 5.]
[[ 0.6 -0.2]
 [-0.2  0.4]]
2 1
[3 6] [3 6]
```

`np.linalg.solve(A, b)` finds the input directly, without building the inverse, and is what you reach for. Asking it to undo the squash fails:

This cell fails on purpose: the lesson reads the error with you.

````python
np.linalg.solve(B, np.array([3, 6]))   # LinAlgError: Singular matrix
````

“Singular” is the word for a matrix with no inverse. The grey machine and a 3 by 3 whose rank is below 3, checked the same way:

````python
G = np.array([[0.299, 0.587, 0.114]])
red, green = np.array([1, 0, 0]), np.array([0, 0.299 / 0.587, 0])
print(G @ red, G @ green)
C = np.array([[1, 0, 1], [0, 1, 1], [1, 1, 2]])
print(np.linalg.matrix_rank(C), np.linalg.matrix_rank(G))
````

## 3.8 A network layer: a matrix, a shift and a bend

A network layer is a matrix, a shift and a bend, and this lesson writes one out with every shape tracked. It also shows why the bend matters, using a problem a single layer cannot solve.

[Read the lesson](https://aicourses.dev/mathematics/matrices/a-layer)

### In NumPy

The layer for one input, then a batch:

````python
import numpy as np

W = np.array([[1, 2], [0, 1], [-1, 1]])
b = np.array([0.5, 0, -1])
x = np.array([2, 1])
h = W @ x + b
print(h, np.maximum(0, h))
X = np.array([[2, 1], [0, 3]])
H = X @ W.T + b
print(H)
print(np.maximum(0, H))
print(X.shape, W.T.shape, H.shape)
````

*The lesson shows:*

```text
[ 4.5  1.  -2. ] [4.5 1.  0. ]
[[ 4.5  1.  -2. ]
 [ 6.5  3.   2. ]]
[[4.5 1.  0. ]
 [6.5 3.  2. ]]
(2, 2) (2, 3) (2, 3)
```

The two rows of `H` are the two paper answers. `np.maximum(0, h)` is the ReLU; it takes the larger of 0 and each entry. Now XOR, with and without the bend:

````python
P = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])    # the four inputs, as rows
W1 = np.array([[1, 1], [1, 1]])
b1 = np.array([0, -1])
w2 = np.array([1, -2])
H = P @ W1.T + b1
print(H @ w2)                    # no bend
print(np.maximum(0, H) @ w2)     # with the bend
````

*The lesson shows:*

```text
[2 1 1 0]
[0 1 1 0]
```

## 3.9 Project: a digit classifier that is one matrix

You stack ten mean digit images into one matrix and classify a digit with a single multiply. It is a complete model made of the module’s pieces, and the one Module 06 will train.

[Read the lesson](https://aicourses.dev/mathematics/matrices/project-digit-templates)

### Ten templates

A **template** is the mean picture of one digit: add up all the training images of a 3 and divide by their number. Stack the ten templates as the rows of a 10 by 64 matrix $\mathbf{W}$. Then the score of a digit image $\mathbf{x}$ (64 numbers) against template $k$ is the dot product of row $k$ with $\mathbf{x}$, and all ten scores are the one product $\mathbf{W}\mathbf{x}$. The label is the digit with the largest score. The setup is the fixed split of the course:

````python
import numpy as np
from sklearn.datasets import load_digits

rng = np.random.default_rng(0)
X, y = load_digits(return_X_y=True)
X = X / 16
order = rng.permutation(len(X))
X, y = X[order], y[order]
Xtr, ytr = X[:1297], y[:1297]
Xte, yte = X[1297:], y[1297:]

W = np.zeros((10, 64))
for k in range(10):
    W[k] = Xtr[ytr == k].mean(axis=0)     # the mean image of digit k
print(W.shape)
````

*The lesson shows:*

```text
(10, 64)
```

Score the first held-out digit with ten dot products in a loop, then the whole test set with one multiply, writing the batch the NumPy way from lesson 8:

````python
x = Xte[0]
loop = [sum(W[k, j] * x[j] for j in range(64)) for k in range(10)]
print(np.round(loop, 3), yte[0])
scores = Xte @ W.T                        # (500, 64) times (64, 10)
print(scores.shape, np.round(scores[0], 3))
pred = scores.argmax(axis=1)
print((pred == yte).mean(), (pred != yte).sum())
````

### What one matrix wins and loses

Module 02’s rule scored 0.992, so one template per digit gives up 0.092. A digit is not one picture. A 1 can be upright or slanted, with or without a foot, and the mean of all of them is a blurred 1 that matches no one’s handwriting well. The rule that compares against 1,297 examples finds the particular 1 that looks like yours. The templates read 1 worst of all the digits, and 7 best:

````python
d = (Xte**2).sum(1)[:, None] - 2 * Xte @ Xtr.T + (Xtr**2).sum(1)
nn = ytr[d.argmin(axis=1)]
print((nn == yte).mean())
print([round(float((pred[yte == k] == k).mean()), 2) for k in range(10)])
````

*The lesson shows:*

```text
[0.98, 0.77, 0.89, 0.87, 0.9, 0.85, 0.98, 1.0, 0.93, 0.84]
```

### Adding the shift

Look for the smallest of these over the ten templates. The first term, $\lVert \mathbf{x} \rVert^2$, is the same for every template, so it cannot change which is smallest and may be dropped. What is left is the smallest of $\lVert \mathbf{w} \rVert^2 - 2\, \mathbf{x} \cdot \mathbf{w}$. Dividing by $-2$ turns smallest into largest, so it is the largest of $\mathbf{x} \cdot \mathbf{w} - \tfrac{1}{2} \lVert \mathbf{w} \rVert^2$. So the rule “nearest template” is the layer $\mathbf{W}\mathbf{x} + \mathbf{b}$ with the shift $b_k = -\tfrac{1}{2} \lVert \mathbf{w}_k \rVert^2$, with no bend, and the largest score wins.

````python
b = -0.5 * (W ** 2).sum(axis=1)
pred_b = (scores + b).argmax(axis=1)
dist = ((W[None] - Xte[:, None]) ** 2).sum(axis=2)    # (500, 10) distances
print((pred_b == yte).mean(), (pred_b == dist.argmin(axis=1)).all())
print((pred != pred_b).sum(), ((pred != yte) & (pred_b == yte)).sum(),
      ((pred == yte) & (pred_b != yte)).sum())
````

*The lesson shows:*

```text
0.898 True
18 7 8
```

### Squash it to rank 1

Squash the ten rows with the projection of Module 02. Take their average, $\mathbf{m}$, and replace each template with its projection on $\mathbf{m}$: $c_k \mathbf{m}$, where $c_k = \mathbf{w}_k \cdot \mathbf{m}$ divided by $\mathbf{m} \cdot \mathbf{m}$. A projection lands on the line along $\mathbf{m}$, so every row is a multiple of the same row and the matrix has rank 1 (lesson 7).

````python
m = W.mean(axis=0)                  # the average template
c = W @ m / (m @ m)                 # how much of m each template holds
W1 = np.outer(c, m)                 # every row is c[k] times m
print(np.linalg.matrix_rank(W1), c.round(3))
pred1 = (Xte @ W1.T).argmax(axis=1)
print((pred1 == yte).mean(), np.unique(pred1), np.bincount(yte))
pred1b = (Xte @ W1.T - 0.5 * (W1 ** 2).sum(axis=1)).argmax(axis=1)
print((pred1b == yte).mean())
````
