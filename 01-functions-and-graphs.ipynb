## 1.2 Straight lines and curves

Slope is the first idea that tells a line from a curve, and it is the seed of the derivative. Here you measure it on the penguin line, then meet a curve whose steepness changes.

[Read the lesson](https://aicourses.dev/mathematics/functions-and-graphs/lines-and-curves)

### The same in NumPy

Python lets you name a small recipe so you can use it again. That is a **function definition**: the word `def`, a name, the inputs in brackets, a colon, then the indented lines that run when you use it, ending with `return`, which hands back the answer. Next comes a **`for` loop**, which runs its indented lines once for each item in a list; here each item is a pair of flipper lengths, and `x1, x2` take the two numbers in turn.

````python
import numpy as np

w, b = 48.9138, 3874.73           # the best line on the twelve

def line(flipper):
    return b + w * (flipper - 198)

for x1, x2 in [(181, 191), (198, 208), (212, 222)]:
    rise = line(x2) - line(x1)
    print(x1, x2, round(rise, 2), round(rise / (x2 - x1), 4))
````

### A curve changes its slope

Now a curve. A car brakes hard, slowing by 8 metres per second every second. From physics, the distance it needs to stop is the speed squared divided by 16 (in metres, with the speed in metres per second). Twice as fast does not take twice the distance. Measure the slope between pairs of speeds, 10 metres per second apart:

````python
def braking(speed):                # metres needed to stop
    return speed ** 2 / 16

for v1, v2 in [(10, 20), (20, 30), (30, 40)]:
    rise = braking(v2) - braking(v1)
    print(v1, v2, rise, rise / (v2 - v1))
````

## 1.3 Chaining machines

A network is a long chain of functions, each feeding the next. This lesson shows how chaining works and why order matters, and it plants the idea behind the chain rule of Module 04.

[Read the lesson](https://aicourses.dev/mathematics/functions-and-graphs/chaining-machines)

### Machines in a row

Python needs two things you may not have met. A **function definition** starts with `def`, a name and the inputs in brackets, and the indented lines under it run when you use it; `return` hands back the answer. A **`for` loop** runs its indented lines once for each item of a list. Here the loop takes the machines in order and feeds each one the previous answer.

````python
import numpy as np

w, b = 48.9138, 3874.73

def line(flipper):                # millimetres in, grams out
    return b + w * (flipper - 198)

def to_kg(grams):
    return grams / 1000

def run_chain(machines, x):
    for m in machines:            # take the machines in order
        x = m(x)                  # feed each one the last answer
    return x

print(line(208), run_chain([line, to_kg], 208))
````

*The lesson shows:*

```text
4363.868 4.363868
```

### A chain of lines is one line

That is bad news for a network built from lines alone: all its layers could be replaced by a single layer. The cure is a **bend** between the lines, a machine that is not straight. A simple one is `np.maximum(0, x)`, which gives back the larger of 0 and its input, so a negative number becomes 0. Chain a line, a bend and a line:

````python
def first(x):  return x - 1
def bend(x):   return np.maximum(0, x)
def second(x): return 3 * x + 2

xs = np.arange(-2, 5)                      # -2, -1, 0, 1, 2, 3, 4
print(xs)
print(run_chain([first, bend, second], xs))
````

*The lesson shows:*

```text
[-2 -1  0  1  2  3  4]
[ 2  2  2  2  5  8 11]
```

### Nudging a chain

Chains are where the chain rule of Module 04 comes from. Nudge the first machine’s input and see how far the last machine’s output moves:

````python
print(run_chain([line, to_kg], 209) - run_chain([line, to_kg], 208))
````

*The lesson shows:*

```text
0.04891379999999934
```

### Sigma is a loop that adds

In Python, `for i in range(1, 5)` runs its indented lines once for each whole number from 1 up to, but not including, 5: so 1, 2, 3, 4.

````python
import numpy as np

a = [1, 2, 3, 6]
total = 0
for i in range(1, 5):          # i = 1, 2, 3, 4, as on paper
    total = total + a[i - 1]   # paper's a_i is Python's a[i - 1]
print(total, sum(a), np.sum(a))
````

### A real formula: mean and spread

Use $a = (1, 2, 3, 6)$, so $H = 4$. The mean is $12 / 4 = 3$. The distances from it are $-2, -1, 0, 3$, their squares $4, 1, 0, 9$ add to 14, and $14 / 4 = 3.5$. The square root is $\sigma = 1.8708$.

````python
a = np.array([1, 2, 3, 6])
mu = 0
for i in range(4):
    mu = mu + a[i] / 4
ss = 0
for i in range(4):
    ss = ss + (a[i] - mu) ** 2
print(mu, round((ss / 4) ** 0.5, 4), np.mean(a), round(np.std(a), 4))
````

### Pi is a loop that multiplies

(in its equation 2, which also conditions on a vector $c$ that summarises the source sentence). Aloud: “the chance of the sentence is the product, for $t$ from 1 to $T$, of the chance of word $t$ given the words before it.” Here $T$ is how many words the sentence has and $y_t$ is the word in position $t$. The bar $\mid$ is read “given”. So each word’s chance is judged knowing the words that came before it, and the sentence’s chance is all those chances multiplied. Take three words with chances 0.5, 0.2 and 0.1: $0.5 \times 0.2 \times 0.1 = 0.01$.

````python
chances = [0.5, 0.2, 0.1]
prod = 1
for c in chances:              # a loop can walk the list directly
    prod = prod * c
print(prod, np.prod(chances))
````

### Subscripts are positions

The same paper uses a sum with weights. Its equation 5 builds a vector $c_i$ for each output word as $c_i = \sum_{j=1}^{T_x} \alpha_{ij} h_j$. Aloud: “$c$ sub $i$ equals the sum, for $j$ from 1 to $T_x$, of alpha sub $i$ $j$ times $h$ sub $j$.” Here $i$ stays fixed while $j$ counts, $T_x$ is the length of the source sentence, $h_j$ describes source word $j$ (in the paper a list of numbers; one number is enough here), and $\alpha_{ij}$ says how much output word $i$ pays attention to source word $j$. Lesson 10 shows how those weights are made. With weights $0.5, 0.3, 0.2$ on three values $10, 20, 30$, the sum is $0.5 \times 10 + 0.3 \times 20 + 0.2 \times 30 = 5 + 6 + 6 = 17$.

````python
alpha = np.array([0.5, 0.3, 0.2])
h = np.array([10, 20, 30])
print(sum(alpha[j] * h[j] for j in range(3)), np.sum(alpha * h))
````

## 1.5 Powers, growth and the number e

Powers and the number e describe growth, and they come back in softmax, in learning-rate schedules and in the logarithms of the next lesson. You meet them from a simple story of repeated multiplying.

[Read the lesson](https://aicourses.dev/mathematics/functions-and-graphs/powers-and-e)

### Powers

The three fit one pattern: each step down in the exponent divides by the base once more, so $2^3 = 8$, $2^2 = 4$, $2^1 = 2$, $2^0 = 1$, $2^{-1} = 0.5$. A rule ties them together: $a^m \times a^n = a^{m+n}$. Multiplying powers of one base adds their exponents. $2^3 \times 2^4$ is $8 \times 16 = 128 = 2^7$.

````python
import numpy as np

print(2 ** 10, 2 ** -3, 4 ** 0.5)
print(round(4000 ** -0.5, 6))
````

*The lesson shows:*

```text
1024 0.125 2.0
0.015811
```

### Growth by a fixed share

Powers are what growth looks like when something rises by the same share each step. A balance of 100 that grows by 5% a year is multiplied by 1.05 each year: 105, then 110.25, then 115.76. After $n$ years it is $100 \times 1.05^n$. The size of the jump grows, because each year’s growth earns growth.

````python
balance = 100
for year in range(10):             # ten turns of the loop
    balance = balance * 1.05
print(round(balance, 2), round(100 * 1.05 ** 10, 2))
````

*The lesson shows:*

```text
162.89 162.89
```

### The number e

Keep splitting the year. In $n$ equal parts, each pays $1/n$ of the 100%, so after the year 1 has become $(1 + 1/n)^n$. For $n = 1$ that is 2, for $n = 2$ it is 2.25. For $n = 12$ (monthly) it is $(1 + 1/12)^{12}$:

````python
for n in [1, 2, 12, 365, 1_000_000]:
    print(n, round((1 + 1 / n) ** n, 5))
print(round(np.e, 5), round(np.exp(1), 5))
````

## 1.6 Logarithms

A logarithm undoes a power and turns multiplying into adding, which is why models work with logs of probabilities. Module 09 depends on this lesson more than on any other in the module.

[Read the lesson](https://aicourses.dev/mathematics/functions-and-graphs/logarithms)

### Which log?

This course writes one of these two every time and never a bare “log”. But papers and code are not so careful. In many papers a plain $\log$ means $\ln$, and in NumPy `np.log` is the natural logarithm, not base 10. The base-2 log is `np.log2`. When you meet a bare `log`, assume the natural one unless the paper says otherwise, and check. The code below is that check for NumPy: `np.log(np.e ** 2)` prints 2.0, so `np.log` asks “what power of $e$?”.

````python
import numpy as np

print(np.log2(8), np.log2(1024), np.log2(1 / 8))
print(np.log(np.e ** 2), round(np.log(9), 4))
````

*The lesson shows:*


### Multiplying becomes adding

Take $a = 2$ and $b = 3$. $\ln 2 = 0.6931$ and $\ln 3 = 1.0986$, which add to 1.7917; and $\ln 6 = 1.7918$. The last digit differs only because the two logs were rounded before they were added, and the code below shows the unrounded values agree.

````python
print(np.log(2) + np.log(3), np.log(6))
print(np.log2(8) + np.log2(16), np.log2(8 * 16))
````

## 1.7 The curves neural networks use

Sigmoid, tanh and ReLU are the bends that let a network do more than a straight line can. You see where each one bends and where it goes flat, which matters again in Module 04.

[Read the lesson](https://aicourses.dev/mathematics/functions-and-graphs/curves-networks-use)

### Sigmoid: a squashed S

By hand at $x = 2$: $e^{-2} = 0.1353$, so $\sigma(2) = 1 / 1.1353 = 0.8808$. Use it in code, for the sigmoid and its two relatives. The **tanh** has the same S shape, but between $-1$ and $1$ (its formula is the sigmoid stretched: $\tanh x = 2 \sigma(2x) - 1$, so $\tanh 2 = 2 \sigma(4) - 1 = 2 \times 0.98201 - 1 = 0.9640$). The **ReLU**, in the next section, is a bend rather than an S.

````python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

x = np.array([-6, -4, -2, 0, 2, 4, 6])
print(x)
print(np.round(sigmoid(x), 4))
print(np.round(np.tanh(x), 4))
print(relu(x))
````

### Where each one goes flat

A curve is flat where its slope is nearly 0. Lesson 2 gave you the tool: rise over run between two inputs. Measure it near the middle, and then near the ends:

````python
def slope(f, x1, x2):
    return (f(x2) - f(x1)) / (x2 - x1)

for name, f in [("sigmoid", sigmoid), ("tanh", np.tanh), ("relu", relu)]:
    print(name, round(slope(f, 0, 2), 4), round(slope(f, 4, 6), 4),
          round(slope(f, -6, -4), 4))
````

### Too big, too small

A float has limits at both ends, and gaps between neighbours that depend on where you are. NumPy’s `np.finfo` reports them for each format. The **float32** format uses 32 bits and **float64**, Python’s own, 64. Machine **epsilon** is the gap between 1 and the next float above it. `np.spacing(x)` gives the gap from `x` to the next float above it. In code, `1e16` means $1 \times 10^{16}$ (the `e` stands for “times ten to the”; it is not the number $e$), so the line `1e16 + 1 == 1e16` asks whether adding 1 to ten million billion changes it.

````python
for ft in (np.float32, np.float64):
    info = np.finfo(ft)
    print(ft.__name__, info.max, info.eps)
print(np.spacing(1.0), np.spacing(1e16), 1e16 + 1 == 1e16)
print(float(np.finfo(np.float16).max))
````

### How models work round it

Softmax (Module 09) turns scores into chances by taking $\exp$ of each and dividing by the sum. For the scores 1000 and 1001 the naive version overflows, and $\text{inf} / \text{inf}$ is `nan`, “not a number”. The fix is to subtract the largest score from every score first. That does not change the answer, because the extra factor appears on the top and on the bottom and cancels, but it keeps every $\exp$ at 1 or below:

````python
z = np.array([1000.0, 1001.0])
with np.errstate(over="ignore", invalid="ignore"):
    print(np.exp(z) / np.exp(z).sum())          # the naive version
e = np.exp(z - z.max())                        # now exp(-1) and exp(0)
print(np.round(e / e.sum(), 5))
````

*The lesson shows:*

```text

[0.26894 0.73106]
```

### Step 3: softmax’s inner sum

From Bahdanau, Cho and Bengio (2014), equation 6: $\alpha_{ij} = \exp(e_{ij}) / \sum_{k=1}^{T_x} \exp(e_{ik})$. Aloud: “alpha sub $i$ $j$ is $\exp$ of $e$ sub $i$ $j$, over the sum, for $k$ from 1 to $T_x$, of $\exp$ of $e$ sub $i$ $k$.” These are the weights of lesson 4’s weighted sum, made from scores. Here $e_{ij}$ is a score (not the number $e$), $T_x$ is how many scores there are in the row, and $\alpha_{ij}$ is the weight made from them. Take one row, $i$ fixed, with three scores $2.0$, $1.0$ and $0.1$. By hand, the three exps are 7.389, 2.718 and 1.105. Their sum is 11.2125 (adding the rounded three gives 11.212; the unrounded values give 11.2125), and so the weights are 0.659, 0.2424 and 0.0986, which add to 1.

````python
e = [2.0, 1.0, 0.1]
bottom = 0
for k in range(3):                     # the sum under the fraction
    bottom = bottom + np.exp(e[k])
alpha = []
for j in range(3):
    alpha.append(np.exp(e[j]) / bottom)
print(round(bottom, 4), np.round(alpha, 4))

e = np.array(e)
print(np.round(np.exp(e) / np.exp(e).sum(), 4))
````

### Step 4: a learning-rate schedule

From the transformer paper (Vaswani and others, 2017), equation 3: $\text{lrate} = d_{\text{model}}^{-0.5} \cdot \min(\text{step}^{-0.5},\ \text{step} \cdot \text{warmup}^{-1.5})$, with warmup 4000 and $d_{\text{model}}$ 512. Aloud: “the learning rate is $d_{\text{model}}$ to the minus a half, times the smaller of two numbers: the step to the minus a half, and the step times warmup to the minus three halves.” The first of the two numbers falls as steps pass, the second rises, and they cross at step 4000: so the rate climbs for 4000 steps, then falls. (The paper writes the step as `step_num` and the warmup as `warmup_steps`.) The peak should be about 0.0007. The `f"{lr:.2e}"` shows a number in scientific notation to two decimals (1.75e-07 is $1.75 \times 10^{-7}$). `min` picks the smaller of two numbers, and `np.minimum` does it for each pair of entries of two arrays.

````python
d_model, warmup = 512, 4000
for step in [1, 1000, 4000, 16000]:
    lr = d_model ** -0.5 * min(step ** -0.5, step * warmup ** -1.5)
    print(step, f"{lr:.2e}")

steps = np.array([1, 1000, 4000, 16000])
print(d_model ** -0.5 * np.minimum(steps ** -0.5, steps * warmup ** -1.5))
````

*The lesson shows:*

```text

1000 1.75e-04
```

### Step 5: a sinusoid

From section 3.5 of the same paper: $PE_{(pos,\,2i)} = \sin(pos / 10000^{2i/d_{\text{model}}})$ and $PE_{(pos,\,2i+1)} = \cos(pos / 10000^{2i/d_{\text{model}}})$. Aloud: “the position signal for position $pos$, in dimension $2i$, is the sine of $pos$ over ten thousand to the power two $i$ over $d$; in dimension $2i + 1$, the cosine of the same.” Take $d_{\text{model}} = 4$, so $i$ is 0 or 1. For $i = 0$ the divisor is $10000^0 = 1$; for $i = 1$ it is $10000^{0.5} = 100$. Each row is $[\sin(pos), \cos(pos), \sin(pos/100), \cos(pos/100)]$. $PE$ is the position signal, a row of $d_{\text{model}}$ numbers for each position. In the code, `d // 2` is $d$ divided by 2 and dropped to a whole number, here 2; `np.zeros(d)` is a row of $d$ zeros; and in the NumPy version `pe[0::2]` means every second entry starting from the first, and `pe[1::2]` every second entry starting from the second.

````python
d = 4
for pos in range(4):
    pe = []
    for i in range(d // 2):
        angle = pos / 10000 ** (2 * i / d)
        pe.append(np.sin(angle))
        pe.append(np.cos(angle))
    print(pos, np.round(pe, 4))

for pos in range(4):
    angle = pos / 10000 ** (2 * np.arange(d // 2) / d)
    pe = np.zeros(d)
    pe[0::2], pe[1::2] = np.sin(angle), np.cos(angle)
    print(pos, np.round(pe, 4))
````
