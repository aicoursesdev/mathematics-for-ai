## 7.2 Events and the rules for and, or, not

A grid of 100 dots is enough to read off the rules for and, or and not. You learn them by counting, so there are no formulas to memorise.

[Read the lesson](https://aicourses.dev/mathematics/probability/events-and-rules)

### In NumPy

Each event is a list of 100 yes-or-no answers, one per email. A loop counts the emails in both; then NumPy does the counting in one line, because `&` means and, `|` means or, `~` means not, and `.sum()` counts the yeses.

````python
import numpy as np

n = np.arange(100)
free = n < 40                  # emails 0 to 39 contain "free"
link = (n >= 25) & (n < 55)    # emails 25 to 54 hold a link
both = 0
for i in range(100):
    if free[i] and link[i]:
        both += 1
print(both)
print((free & link).sum(), (free | link).sum(), (~free).sum())
````

*The lesson shows:*

```text
15
15 55 60
```

The overlap is emails 25 to 39, which is 15, so the code and the grid agree. A mean of yes-or-no answers is the share of yeses, which is the chance:

````python
print((free & link).mean(), (free | link).mean(), (~free).mean())
````

*The lesson shows:*

```text
0.15 0.55 0.6
```

## 7.3 Conditional probability

To condition on something is to look only at the cases where it happened. This lesson uses the grid to work out the share inside them, the step before Bayes’ rule.

[Read the lesson](https://aicourses.dev/mathematics/probability/conditional-probability)

### In NumPy

A loop counts the emails that fit both, and divides by the number that fit the condition. NumPy does the zooming by indexing: `free[link]` is the list of “free” answers of the emails that have a link, and its mean is the share of yeses among them.

````python
import numpy as np

n = np.arange(100)
free = n < 40
link = (n >= 25) & (n < 55)
both = kept = 0
for i in range(100):
    if link[i]:            # zoom in: skip emails without a link
        kept += 1
        both += free[i]
print(both, kept, both / kept)
print(free[link].mean(), link[free].mean(), free.mean())
````

*The lesson shows:*

```text
15 30 0.5
0.5 0.375 0.4
```

## 7.4 Bayes’ rule and base rates

Bayes’ rule turns a test result back into a chance, and the base rate decides the answer. You work it from counts of 1,000 people, and the formula comes last.

[Read the lesson](https://aicourses.dev/mathematics/probability/bayes-rule)

### In NumPy

The first block is the counting. The second is the formula, written as a small function so you can change the base rate.

````python
import numpy as np

people, sick, hits, false_pos = 1000, 10, 8, 95
positives = hits + false_pos
print(positives, round(hits / positives, 4))
````

*The lesson shows:*

```text

```

````python
def posterior(base, hit, false_rate):
    p_pos = hit * base + false_rate * (1 - base)
    return hit * base / p_pos

for base in [0.001, 0.01, 0.1, 0.5]:
    print(base, round(posterior(base, 0.8, 95 / 990), 3))
````

[Read the lesson](https://aicourses.dev/mathematics/probability/expected-value-and-variance)

A loop adds each value times its chance; NumPy does the same with one product and a sum.

````python
import numpy as np

rng = np.random.default_rng(0)
x = np.arange(1, 7)
p = np.full(6, 1 / 6)
mean = 0
for xi, pi in zip(x, p):
    mean += xi * pi
print(round(mean, 4), (x * p).sum().round(4))
var = (p * (x - mean) ** 2).sum()
print(round(var, 4), round(var ** 0.5, 3))
````

## 7.7 Distributions and sampling

A distribution is a shape with knobs, and drawing from one is how a model picks its next word. You meet the four you need and draw from each with a seeded generator.

[Read the lesson](https://aicourses.dev/mathematics/probability/distributions)

### In NumPy

Each draw comes from `rng`, the seeded generator. A Bernoulli draw is a uniform draw compared with $p$; the sampler you build in the project uses a version of the same trick.

````python
import numpy as np

rng = np.random.default_rng(0)
table = np.array([0.40, 0.25, 0.15, 0.10, 0.10])
tokens = rng.choice(5, size=10_000, p=table)  # categorical
wait = rng.uniform(2, 5, size=10_000)       # uniform from 2 to 5
flips = rng.random(10_000) < 0.3            # Bernoulli, p = 0.3
print(np.bincount(tokens) / 10_000)
print(round(wait.mean(), 3), wait.min() > 2, wait.max() < 5)
print(flips.mean())
````

Every number printed here is **simulated**: it comes from 10,000 draws with seed 0, so it is close to the chance but not exactly it. The five token shares are near the table, the uniform draws average near 3.5 and the Bernoulli share is near 0.3. Now the normal:

````python
z = rng.normal(10, 2, size=10_000)   # mean 10, spread 2
print(round(z.mean(), 3), round(z.std(), 3))
for m in [1, 2, 3]:
    print(m, (abs(z - 10) < m * 2).mean())
````

## 7.8 Many draws: averages settle down

Averages of more draws wander less, and sums of many small things make a bell. This lesson gives the rule and shows it, which is what makes sampling checks in the project possible.

[Read the lesson](https://aicourses.dev/mathematics/probability/many-draws)

### The average settles

The **law of large numbers** says that the average of many independent draws gets close to the expected value as the number of draws grows. For a coin, the share of heads gets close to its chance. It is a statement about the average (the share), not the count: the count of heads can drift further from half the flips even as the share closes in. The next section says how big the wander is.

````python
import numpy as np

rng = np.random.default_rng(0)
flips = rng.random(10_000) < 0.5
for n in [10, 100, 1_000, 10_000]:
    print(n, flips[:n].mean())
````

*The lesson shows:*

```text

100 0.44
1000 0.473
10000 0.499
```

The settling is by **dilution**, not by correction. After a run of six heads, the coin is no more likely to give tails. The streak stays in the count but becomes a small part of it. You can check that with a long simulated run:

````python
f = rng.random(200_000) < 0.5
five = f[:-5] & f[1:-4] & f[2:-3] & f[3:-2] & f[4:-1]
print(five.sum(), f[5:][five].mean())
````

### Sums of many small things make a bell

There is a second fact, and it is why the numbers 0.68 and 0.95 of lesson 7 keep appearing. The **central limit theorem** says that the average of many independent draws from the same shape, almost any shape with a finite spread, comes closer and closer to the bell shape of a normal distribution, with the spread $\sigma / \sqrt{n}$. For a finite $n$ the bell is only approximate. This can be proved; here you check it. Take the made-up table of token lengths from lesson 6: lengths 3, 1, 3, 3 and 8 with chances 0.40, 0.25, 0.15, 0.10 and 0.10. It is lopsided, with one long token. Its mean is 3.0 and its spread is about 1.871. Average 30 draws and the spread should be $1.871 / \sqrt{30}$, about 0.342.

````python
length = np.array([3, 1, 3, 3, 8])
chance = np.array([0.40, 0.25, 0.15, 0.10, 0.10])
for n in [1, 30]:
    avg = rng.choice(length, size=(10_000, n), p=chance).mean(axis=1)
    se = 1.871 / np.sqrt(n)
    print(n, round(avg.std(), 3), round(se, 3),
          (abs(avg - 3) < se).mean(), (abs(avg - 3) < 2 * se).mean())
````

30 0.34 0.342 0.6997 0.9584

### Step 2: walk the running total

Lay the chances end to end on a line from 0 to 1. “the” owns 0 to 0.30, “a” owns 0.30 to 0.50, and so on. Draw one uniform number $u$ between 0 and 1, and the token whose stretch contains it is the pick. Its chance of being picked is the width of its stretch, which is its chance in the table. The running total is what `np.cumsum` gives.

````python
import numpy as np

rng = np.random.default_rng(0)
tokens = ["the", "a", "cat", "sat", "on", "mat", "."]
table = np.array([0.30, 0.20, 0.15, 0.10, 0.10, 0.08, 0.07])
print(table.sum().round(10), len(tokens))
cum = np.cumsum(table)
cum[-1] = 1.0                    # rounding can leave the total just under 1
print(cum.round(2))

def walk(u):
    for i, total in enumerate(cum):
        if u < total:
            return i
print(tokens[walk(0.42)], tokens[walk(0.95)])
````

### Step 3: draw 10,000 times

Draw 10,000 uniform numbers at once, one for each draw. The loop walks the running total for each; `np.searchsorted` finds the same stretch for all of them in one call. The two must agree. Then tally.

````python
u = rng.random(10_000)
by_loop = np.array([walk(x) for x in u])
draws = np.searchsorted(cum, u, side="right")
print((by_loop == draws).all())
counts = np.bincount(draws, minlength=7)
print(counts)
print(counts / 10_000)
````

### Step 4: is it as close as chance allows?

Lesson 8 gives the yardstick. A token with chance $p$, drawn $n$ times, has a share that wanders by about $\sqrt{p(1-p)/n}$. For “the”, $p = 0.30$ and $n = 10{,}000$: $\sqrt{0.21/10000} = 0.00458$. Measure each token’s gap, share minus chance, in these wanders. About 95 in 100 gaps should be within 2 wanders, and nearly all within 3. With 7 tokens, one gap beyond 2 now and then is ordinary. A gap beyond 3, or two or more beyond 2, is a warning.

````python
def verdict(draws, n=10_000):
    shares = np.bincount(draws, minlength=7) / n
    se = np.sqrt(table * (1 - table) / n)
    z = (shares - table) / se
    return z.round(2), int((abs(z) > 2).sum()), float(abs(z).max().round(2))

print(verdict(draws))
````
