## 4.2 The slope at a point

You find the slope at a single point by shrinking a gap until the answer settles. A table of shrinking steps makes the derivative something you can compute and check.

[Read the lesson](https://aicourses.dev/mathematics/derivatives/slope-at-a-point)

### In NumPy

The same table in a loop. The function takes $x$ and returns $x^2$, and the loop does the division by the step.

````python
import numpy as np

def f(x):
    return x ** 2

for h in [1, 0.1, 0.01, 0.001]:
    print(h, (f(3 + h) - f(3)) / h)
````

The long tails of digits are the rounding of Module 01.9: the numbers are right to about 12 places. NumPy does every step at once when `h` is an array, and the same code gives the slope at $-2$:

````python
h = np.array([1, 0.1, 0.01, 0.001])
print((f(3 + h) - f(3)) / h)
print((f(-2 + h) - f(-2)) / h)
````

## 4.3 Up close, every curve is nearly a line

Up close, every curve looks like a line, and that single idea justifies each step of gradient descent. It is the lesson to come back to when a later formula feels magical.

[Read the lesson](https://aicourses.dev/mathematics/derivatives/nearly-a-line)

### In NumPy

Here is the table of guesses and gaps, with the slope 6 typed in. `round` keeps the rounding noise of Module 01.9 out of view.

````python
import numpy as np

def f(x):
    return x ** 2

slope = 6                        # the slope of x**2 at 3
for h in [1, 0.1, 0.01]:
    guess = f(3) + slope * h     # the straight-line guess
    print(h, guess, round(f(3 + h), 4), round(f(3 + h) - guess, 4))
````

*The lesson shows:*

```text

0.1 9.6 9.61 0.01
0.01 9.06 9.0601 0.0001
```

### Why a step against the slope works

A square is never negative, so for a small enough $\eta$ the loss falls, whatever the slope’s sign. Check it at $w = 1$ with $\eta = 0.05$: the change in $w$ is $-0.05 \times (-6) = 0.3$, and the guess for the new loss is $10 - 0.05 \times 36 = 8.2$.

````python
def loss(w):
    return (w - 4) ** 2 + 1

s = -6                           # the slope of the loss at w = 1
print(loss(1), round((loss(1.001) - loss(1)) / 0.001, 3))
step = -0.05 * s                 # a step against the slope
print(round(step, 2), round(loss(1) + s * step, 2), round(loss(1 + step), 2))
````

```text

0.3 8.2 8.29
```

### Checking the rules in NumPy

A nudge of $10^{-6}$ gives the slope to about six digits, which is plenty to check a hand answer. The function `slope` takes any function.

## 4.6 Derivatives of the curves AI uses

### Exponentials and logarithms

At $x = 2$ that is $\frac12$ and at $x = 4$ it is $\frac14$. The picture fits what you know of $\ln x$: near 0 it climbs steeply, and for large $x$ it flattens, so its slope shrinks as $x$ grows. Both slopes can be proved (the one for $\ln x$, which needs $x > 0$, follows from the chain rule). Here you check them with a nudge:

````python
import numpy as np

def slope(f, x, h=1e-6):
    return (f(x + h) - f(x)) / h

for x in [0, 1, 2]:
    print(x, round(slope(np.exp, x), 3), round(np.exp(x), 3))
for x in [0.5, 2, 4]:
    print(x, round(slope(np.log, x), 4), 1 / x)
````

### Tanh and ReLU

ReLU, $\max(0, x)$, is the simplest. Its slope is 0 for $x$ below 0, where the curve lies flat on the axis, and 1 for $x$ above 0, where it is the line $y = x$. At exactly 0 there is a corner with no single slope, and a program has to choose a number for that one point. The slopes of the curves, checked with the code you now know:

````python
s = lambda x: 1 / (1 + np.exp(-x))
for x in [0, 1, 2, 5, 10]:
    print(x, round(s(x), 4), round(s(x) * (1 - s(x)), 6), round(slope(s, x), 6))
for x in [0, 1, 2, 5]:
    print(x, round(1 - np.tanh(x) ** 2, 4), round(slope(np.tanh, x), 4))
relu = lambda x: np.maximum(0, x)
print([round(slope(relu, x), 3) for x in [-2, -0.5, 0.5, 2]])
````

## 4.7 Curvature: the second derivative

The second derivative says how fast the slope itself changes. It gives the curvature that will set the limit on the learning rate in lesson 10.

[Read the lesson](https://aicourses.dev/mathematics/derivatives/curvature)

### In NumPy

A function that returns the slope of another function lets you apply it twice. The step is 0.001, which is fine here and is the subject of lesson 9.

````python
import numpy as np

def d(f, h=1e-3):
    return lambda x: (f(x + h) - f(x)) / h      # the slope, as a function

f1, f2, f3 = lambda x: x ** 2, lambda x: -x ** 2, lambda x: 5 * x ** 2
for f in [f1, f2, f3]:
    print(round(d(f)(3), 1), round(d(d(f))(3), 1))
g = lambda x: x ** 3 - 6 * x
for x in [1, -1]:
    print(x, round(d(g)(x), 1), round(d(d(g))(x), 1))
````

## 4.8 Finding the lowest point

At the lowest point the slope is zero, and for the one-knob penguin fit that gives the best answer in one line of algebra. It is the exact answer your search and your descent will be measured against.

[Read the lesson](https://aicourses.dev/mathematics/derivatives/the-lowest-point)

### One knob, on purpose

Here $\hat{y}$ (“y hat”) is the model’s guess for $y$, and $w$ is the one knob, in kilograms per centimetre. Why centre? The best line must pass through the middle of the cloud, and after centring the middle is at $(0, 0)$. So the line’s intercept is 0 and only its slope is left to learn. The best slope does not change by centring.

````python
import numpy as np

flipper = np.array([181, 183, 185, 192, 193, 193,
                    197, 198, 205, 208, 216, 222], dtype=float)     # mm
mass = np.array([3625, 3550, 3150, 2700, 2925, 3650,
                 3750, 4400, 4300, 4200, 4750, 5350], dtype=float)  # g
x = (flipper - flipper.mean()) / 10    # cm above the average
y = (mass - mass.mean()) / 1000        # kg above the average
print(x.round(3))
print(y.round(4))
````

## 4.9 Checking a derivative with numbers

A derivative you cannot check is a derivative you cannot trust, so this lesson gives you a way to check any of them with numbers. It also shows why a tinier step can make the check worse.

[Read the lesson](https://aicourses.dev/mathematics/derivatives/checking-with-numbers)

### The central difference

````python
import numpy as np

f = lambda x: x ** 2
print((f(3 + 1) - f(3)) / 1, (f(3 + 1) - f(3 - 1)) / 2)
g = lambda x: x ** 3
print((g(2.1) - g(1.9)) / 0.2, (g(2.1) - g(2)) / 0.1)
````

*The lesson shows:*

```text

```

### A smaller step is not always better

Here is the error of both difference formulas for $e^x$ at 1, each estimate minus the true slope $e = 2.71828\ldots$. The table has three columns: the step $h$, the error of the one-sided nudge and the error of the central difference. An error is printed in computer notation: `1.4e-01` means $1.4 \times 10^{-1}$, which is 0.14, and `5.9e-11` means $0.000000000059$. The bigger the number after `e-`, the smaller the error, and a minus sign in front only says the estimate came out too low. What to look for is the row in each column where the size of the error is smallest.

````python
exact = np.exp(1)
for h in [1e-1, 1e-3, 1e-5, 1e-8, 1e-10, 1e-12, 1e-14, 1e-16]:
    fw = (np.exp(1 + h) - np.exp(1)) / h
    ce = (np.exp(1 + h) - np.exp(1 - h)) / (2 * h)
    print("%.0e  %9.1e  %9.1e" % (h, fw - exact, ce - exact))
````

### A check you can use

For a hand derivative, use the central difference with $h$ about $10^{-5}$: the table says it is accurate there to ten digits, with plenty of room either side. Compare a hand derivative with it, and call it right when the two agree to seven or eight digits. The test is the **relative difference**: the gap between the two numbers divided by the size of the central difference, so 0.5 means they disagree by half. Here the hand derivative of sigmoid(2x) from lesson 6, and the same with the inner 2 forgotten:

````python
def check(f, df, x, h=1e-5):
    num = (f(x + h) - f(x - h)) / (2 * h)
    return abs(df(x) - num) / abs(num)       # the relative difference

s = lambda x: 1 / (1 + np.exp(-x))
f = lambda x: s(2 * x)
good = lambda x: 2 * s(2 * x) * (1 - s(2 * x))
bad = lambda x: s(2 * x) * (1 - s(2 * x))    # forgot the inner 2
print(check(f, good, 1) < 1e-8, round(check(f, bad, 1), 3))
````


## 4.10 Walking downhill: gradient descent with one knob

Now the slope steers a walk downhill, with one knob and a learning rate. You see it crawl, converge, bounce and fly apart, which is what the rest of Module 06 tries to tame.

[Read the lesson](https://aicourses.dev/mathematics/derivatives/walking-downhill)

### Three runs on the penguins

Here is the one-knob problem of lesson 8 again, with its loss, its slope $L'(w) = 2 \cdot \text{mean}\big((w x - y) x\big)$ and its curvature 3.0304.

````python
import numpy as np

flipper = np.array([181, 183, 185, 192, 193, 193,
                    197, 198, 205, 208, 216, 222], dtype=float)     # mm
mass = np.array([3625, 3550, 3150, 2700, 2925, 3650,
                 3750, 4400, 4300, 4200, 4750, 5350], dtype=float)  # g
x = (flipper - flipper.mean()) / 10    # cm above the average
y = (mass - mass.mean()) / 1000        # kg above the average
L = lambda w: np.mean((w * x - y) ** 2)
slope = lambda w: 2 * np.mean((w * x - y) * x)
curv = 2 * np.mean(x * x)
best = np.sum(x * y) / np.sum(x * x)
print(round(L(0), 4), round(slope(0), 4), round(curv, 4), round(2 / curv, 4))
````

*The lesson shows:*

```text
0.5454 -1.4823 3.0304 0.66
```

Now the loop, started at $w = 0$, run for five steps at three learning rates:

````python
def descend(eta, steps=5, w=0.0):
    ws = [w]
    for _ in range(steps):
        w = w - eta * slope(w)
        ws.append(w)
    return np.array(ws)

for eta in [0.02, 0.3, 0.7]:
    ws = descend(eta)
    print(eta, ws.round(4), round(1 - eta * curv, 4))
    print("  loss", np.array([L(w) for w in ws]).round(4))
````

```text
  loss [0.5454 0.1858 0.1829 0.1828 0.1828 0.1828]
```

### The speed limit

The distance shrinks only if the factor lies between $-1$ and 1. The factor $1 - \eta c$ is below 1 whenever $\eta$ is above 0, and it is above $-1$ when $\eta c$ is below 2, that is when $\eta < \frac{2}{c}$. So the walk converges when $0 < \eta < \frac{2}{c}$. For this loss that limit is $\frac{2}{3.0304} = 0.6600$. Exactly there the factor is $-1$: the step lands on the mirror point of lesson 7, as far from the bottom as you began but on the other side, and the walk bounces forever between two points. Above the limit the factor is below $-1$, so each step lands beyond the mirror point, farther out than the last. The table below runs 20 steps from $w = 0$ and shows the distance left:

````python
for eta in [0.60, 0.65, 0.66, 0.67, 0.70]:
    w = descend(eta, 20)[-1]
    print(eta, round(1 - eta * curv, 4), round(abs(w - best), 4))
````

### Step 2: derive the slope

Write that as `slope(w)`. At $w = 0$ it is $-2 \cdot \text{mean}(x y) = -1.4823$, and hand-checking it that way against lesson 8’s $B$ is a good first test.

````python
import numpy as np

flipper = np.array([181, 183, 185, 192, 193, 193,
                    197, 198, 205, 208, 216, 222], dtype=float)     # mm
mass = np.array([3625, 3550, 3150, 2700, 2925, 3650,
                 3750, 4400, 4300, 4200, 4750, 5350], dtype=float)  # g
x = (flipper - flipper.mean()) / 10    # cm above the average
y = (mass - mass.mean()) / 1000        # kg above the average
L = lambda w: np.mean((w * x - y) ** 2)
slope = lambda w: 2 * np.mean((w * x - y) * x)      # your derivation
best = np.sum(x * y) / np.sum(x * x)                # lesson 8's exact answer
print(round(L(0), 4), round(slope(0), 4), round(best, 4))
````

### Step 4: descend

Now let the knob learn. Start at $w = 0$, use $\eta = 0.3$, the learning rate that worked in lesson 10, step against the slope, and stop when the slope is smaller than $10^{-6}$. Count the slope evaluations as you go.

````python
w, n = 0.0, 1                  # n counts slope evaluations
s = slope(w)
while abs(s) > 1e-6:
    w -= 0.3 * s
    s = slope(w)
    n += 1
print(n, round(w, 5), round(best, 5), round(L(w), 5))
print(abs(w - best) < 1e-6)
````

### Step 5: compare with the search

Last, the loop of Module 00 on the same problem: try $w$ a step up and a step down, keep the better, stop when neither helps. With one knob it makes two tries per step. Use a step of 0.01.

````python
w, n, step = 0.0, 1, 0.01      # n counts error evaluations
now = L(w)
while True:
    tries = [w + step, w - step]
    errs = [L(t) for t in tries]
    n += 2
    i = int(np.argmin(errs))
    if errs[i] >= now:
        break
    w, now = tries[i], errs[i]
print(n, round(w, 3), round(abs(w - best), 5))
````
