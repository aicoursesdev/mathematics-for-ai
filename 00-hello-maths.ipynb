## 0.5 Error as one number

To turn a knob in the right direction you need to say how wrong the model is with one number. This lesson builds that number, the mean squared error, and shows why it is built the way it is.

[Read the lesson](https://aicourses.dev/mathematics/hello-maths/error-as-one-number)

### By hand, for three penguins

with the unit grams squared, since the misses are in grams. The mean of squares comes out in a strange unit, so people also take its square root, which is back in grams and reads as a typical miss. Next, the same in code.

````python
import numpy as np

flipper = np.array([181, 183, 185, 192, 193, 193,
                    197, 198, 205, 208, 216, 222])
mass = np.array([3625, 3550, 3150, 2700, 2925, 3650,
                 3750, 4400, 4300, 4200, 4750, 5350])
x = flipper - 198
b, w = 3940, 30
for i in [0, 7, 11]:
    miss = mass[i] - (b + w * x[i])
    print(miss, miss ** 2)
````

*The lesson shows:*

```text
195 38025
460 211600
690 476100
```

### All twelve, and the best line

For all twelve birds, first with a loop, then with NumPy’s mean.

````python
miss = mass - (b + w * x)
total = 0
for m in miss:
    total = total + m ** 2
print(round(total / 12, 1))
print(round(np.mean(miss ** 2), 1))
````

*The lesson shows:*

```text
241945.8
241945.8
```

The loop adds up the squares and divides by 12; `np.mean(miss ** 2)` does the same on the whole array. `round(…, 1)` keeps one decimal place. Both give 241945.8 for this line. Now three lines side by side, as a function you can call with any two knobs.

````python
def mse(w, b):
    return np.mean((mass - (b + w * x)) ** 2)

for w, b in [(0, 4000), (30, 3940), (50, 3900)]:
    print(w, b, round(mse(w, b), 1))
````

*The lesson shows:*

```text
0 4000 564270.8
30 3940 241945.8
50 3900 183645.8
```

`def` defines a function, a named recipe with inputs (here w and b) that gives back (`return`) an answer. In the `for` line, each pair in the list is split into `w` and `b` in turn. The flat line at 4000 g scores 564271, and each better guess scores less. Is there a best line? One routine finds it (`np.polyfit` returns the two knobs that give the lowest error; the 1 asks for a straight line). Module 04 works out the one-knob case by hand, and Module 10 the general one:

````python
w, b = np.polyfit(x, mass, 1)
print(round(w, 4), round(b, 2), round(mse(w, b), 1))
print(round(np.sqrt(mse(w, b)), 1))
````

*The lesson shows:*

```text
48.9138 3874.73 182842.1
427.6
```

## 0.6 Learning is turning the knobs

With a score for the error in hand, learning becomes a loop: nudge a knob, keep the nudge if the error falls, repeat. The loop works but is slow, and seeing why is what motivates calculus in Module 04.

[Read the lesson](https://aicourses.dev/mathematics/hello-maths/turning-the-knobs)

### The loop

````python
import numpy as np

flipper = np.array([181, 183, 185, 192, 193, 193,
                    197, 198, 205, 208, 216, 222])
mass = np.array([3625, 3550, 3150, 2700, 2925, 3650,
                 3750, 4400, 4300, 4200, 4750, 5350])
x = flipper - 198

def mse(w, b):
    return np.mean((mass - (b + w * x)) ** 2)

w, b = 0, 4000
tries = [(w + 1, b), (w - 1, b), (w, b + 10), (w, b - 10)]
for tw, tb in tries:
    print(tw, tb, round(mse(tw, tb), 1))
````

*The lesson shows:*

```text
1 4000 549530.8
-1 4000 579314.1
0 4010 567120.8
0 3990 561620.8
```

### Running it to the end

Now the whole loop in code. It repeats the four tries and the keep-the-best step, 62 times, until no try helps.

````python
w, b, history = 0, 4000, [mse(0, 4000)]
while True:
    tries = [(w + 1, b), (w - 1, b), (w, b + 10), (w, b - 10)]
    errors = [mse(tw, tb) for tw, tb in tries]
    best = int(np.argmin(errors))
    if errors[best] >= mse(w, b):
        break
    w, b = tries[best]
    history.append(errors[best])
print(len(history) - 1, w, b, round(mse(w, b), 1))
print([round(h) for h in history[::10]])
````

*The lesson shows:*

```text
62 49 3870 182865.8
[564271, 430512, 327071, 253946, 211138, 190923, 182999]
```

### When the loop stalls

One more thing, and it must be said plainly. The loop above worked because flipper length was counted from 198 mm. Count it from zero, as raw millimetres, and the same loop, from w = 0 and b = 0 with steps of 1 and 100, goes wrong.

````python
def nudge(xs, w, b, sw, sb):
    err = lambda w, b: np.mean((mass - (b + w * xs)) ** 2)
    steps = 0
    while True:
        tries = [(w + sw, b), (w - sw, b), (w, b + sb), (w, b - sb)]
        errors = [err(tw, tb) for tw, tb in tries]
        best = int(np.argmin(errors))
        if errors[best] >= err(w, b):
            return steps, w, b, round(float(err(w, b)), 1)
        w, b = tries[best]
        steps = steps + 1

print(nudge(x, 0, 4000, 1, 10))
print(nudge(flipper, 0, 0, 1, 100))
````

*The lesson shows:*

```text
(62, 49, 3870, 182865.8)
(21, 20, -100, 309570.8)
```
